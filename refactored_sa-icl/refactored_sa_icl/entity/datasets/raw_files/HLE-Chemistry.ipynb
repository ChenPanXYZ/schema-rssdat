```python
import numpy as np
import pandas as pd
import string
df = pd.read_csv('HLE_FULL.csv')
print((df['category'].value_counts()))

df = df[(df['answer_type'] == "multipleChoice") & (df['category'] == "Biology/Medicine") & (df['image'].isna()) & (df['rationale_image'].isna())]

print(len(df))
rows = []
for i, row in df.iterrows():
    question = row['question']
    answer_idx = question.find('Answer Choices:')
    question = question[:answer_idx].strip()

    # Parse the answer
    answers = row['question'].split('Answer Choices:')[1].split('\n')
    candidates = [answer[3:] for answer in answers if answer.strip() != '']
    
    label = row['answer'].lower()
    # convert it to an index, it is A, B, C...
    label_idx = string.ascii_uppercase.index(label.upper())
    
    
    correct_answer = row['answer'].lower()
    
    rows.append({
        "Question": question,
        "Correct Answer": candidates[label_idx],
        "Answer Choices": candidates,
        "Explanation": row['rationale'],
    })
    
df = pd.DataFrame(rows)
df.to_csv("HLE_Biology.csv", index=False)
df
```
